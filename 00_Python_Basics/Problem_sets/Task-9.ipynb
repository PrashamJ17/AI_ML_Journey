### `Problem-1:` Class inheritence

Create a **Bus** child class that inherits from the Vehicle class. The default fare charge of any vehicle is seating capacity * 100. If Vehicle is Bus instance, we need to add an extra 10% on full fare as a maintenance charge. So total fare for bus instance will become the final amount = total fare + 10% of the total fare.

Note: The bus seating capacity is 50. so the final fare amount should be 5500. You need to override the fare() method of a Vehicle class in Bus class.

````python
# Write code here
class Vehicle :
    def __init__(self,seats):
        self.seats = seats

    def fare(self):
        return self.seats * 100

class Bus(Vehicle): 
    def fare(self):
        return self.seats*100 + 0.1*self.seats*100
````

````python
bus = Bus(50)
v1 = Vehicle(100)
print(bus.fare())
print(v1.fare())
````

### `Problem-3:` Write a program that has a class Point. Define another class Location which has two objects (Location & Destination) of class Point. Also define a function in Location that prints the reflection of Destination on the x axis.

````python
class Point :

    def __init__(self,x,y):
        self.x = x
        self.y = y

    def show_point(self):
        print(f"{self.x},{self.y}")

class Location(Point) :

    def __init__(self,x1,y1,x2,y2):
        self.source = Point(x1,y1)
        self.destination = Point(x2,y2)
    
    def refl_dest_x(self):
        refl = Point(self.destination.x,-self.destination.y)
        return refl.show_point()
````

````python
L = Location(10,20,30,40)
L.refl_dest_x()
L.source.show_point()
````

### `Problem-4:` Write a program that has an abstract class Polygon. Derive two classes Rectangle and Triamgle from Polygon and write methods to get the details of their dimensions and hence calculate the area.

````python
from abc import ABC,abstractmethod

class Polygon(ABC):

    @abstractmethod
    def get_dim(self):
        pass

    @abstractmethod
    def get_area(self):
        pass


class Rectangle(Polygon):

    def get_dim(self,l,b):
        self.len = l 
        self.bre = b
        print(f"Lenght - {self.len} , Breadth - {self.bre}")
    
    def get_area(self):
        return self.len * self.bre

class Triangle(Polygon):

    def get_dim(self,l,h):
        self.len = l 
        self.hei = h
        print(f"Lenght - {self.len} , Height - {self.hei}")
    
    def get_area(self):
        return self.len * self.hei * 0.5
    
````

````python
# R = Rectangle(10,20)
# R.area_r()

# T= Triangle(3,4,5)
# T.area_t()

r = Rectangle()
t = Triangle()

r.get_dim(10,20)
t.get_dim(3,4)

print(r.get_area())
print(t.get_area())
````

### `Problem-5:` Write a program with class Bill. The users have the option to pay the bill either by cheque or by cash. Use the inheritance to model this situation.

````python
# Parent class - Bill
class Bill :
    def __init__(self,items,price):
        self.items = items
        self.price = price
        self.total = 0

        for i in self.price:
            self.total += i

    def __str__(self):
        return f"Total Amount to be paid - {self.total}"

class Cheque(Bill):
    def __init__(self,items,price,name,acc_no):
        super().__init__(items,price)
        self.name = name
        self.acc_no = acc_no
    
    def payment(self):
        print(super().__str__())
        print(f"Processing Payment - \nAccount No. - {self.acc_no}\nName - {self.name}")
        print(f"Payment Successfull - {self.total} Paid")

class Cash(Bill):
    def __init__(self,items,price,deno,no_of_notes):
        super().__init__(items,price)
        self.deno = deno
        self.no_of_notes = no_of_notes
    
    def payment(self):
        print(super().__str__())
        for i in range(len(self.deno)) :
            print(f"{self.deno[i]} * {self.no_of_notes[i]} = {self.deno[i]* self.no_of_notes[i]}")
 
````

````python
items = ["External Hard Disk", "RAM", "Printer", "Pen Drive"]
price = [5000, 2000, 6000, 800]

deno = [10, 20, 50, 100, 500, 2000]
value = [1, 1, 1, 20, 4, 5]

pay = Cash(items,price,deno,value)
pay.payment()

pay2 = Cheque(items,price,'Prasham Jain',89289910)
pay2.payment()
````
