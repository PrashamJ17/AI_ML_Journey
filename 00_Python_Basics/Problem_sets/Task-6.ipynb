### **`Problem 3:`** Write a Python function that accepts a string and calculate the number of upper case letters and lower case letters.

```
Sample String : 'CampusX is an Online Mentorship Program fOr EnginEering studentS.'
Expected Output :
No. of Upper case characters :  9
No. of Lower case Characters :  47
```

````python
# Write code here
def q3(S):
    u_count = 0
    l_count = 0
    for i in S:
        if i.islower():
            l_count += 1
        elif i.isupper():
            u_count += 1
    print('No. of upper case charcters :',u_count)
    print('No. of lower case charcters :',l_count)
````

````python
q3('CampusX is an Online Mentorship Program fOr EnginEering studentS.')
````

### **`Problem 4:`** Write a Python program to print the even numbers from a given list.
```
Sample List : [1, 2, 3, 4, 5, 6, 7, 8, 9]
Expected Result : [2, 4, 6, 8]
```

````python
# Write code here
def q4(L):
    out=[]
    for i in L :
        if i%2==0:
            out.append(i)
    out1 = [i for i in L if i%2==0]
    a = list(filter(lambda x: x%2==0,L))
    print(a)
    print(out1)
    return out
````

````python
print(q4([1, 2, 3, 4, 5, 6, 7, 8, 9]))
````

```

```
Output

````python
# Write code here
def q7(S):
    L = list(set(S.split()))
    c = []
    for i in L :
        c.append((S.count(i),i))
    a = max(c)
    print(a[1],"->",a[0])
````

`Problem-9` Write a python function that accepts a list of 2D co-ordinates and a query point, and then finds the the co-ordinate which is closest in terms of distance from the query point.

```
List of Coordinates
[(1,1),(2,2),(3,3),(4,4)]
Query Point
(0,0)
```

```
Output
Nearest to (0,0) is (1,1)
```

````python
# Write code here
def q9(L,Q):
    x,y = Q[0],Q[1]
    ans = [((x-i)**2 + (y-j)**2)**0.5 for i,j in L]
    print(min(list(zip(ans,L)))[1])

````

````python
q9([(1,4),(2,-2),(3,3)],(0,0))
````

https://en.wikipedia.org/wiki/Bag-of-words_model

````python
# Write code here
def q10(L):
    L1 = set()
    for i in L :
        for j in i.split():
            for k in j :
                n = ''
                if k=='.':
                    n = j[:-1]
                else :
                    n = j
            L1.add(n)
    print(L1)
    result = []
    for i in L:
        s = []
        for j in L1:
            s.append(i.count(j))
        result.append(s)
    print(result)
        


            
        


        
````
